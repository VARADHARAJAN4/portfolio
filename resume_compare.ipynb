```python
# Portfolio vs Resume Comparison

This notebook loads the portfolio HTML and available resume documents, then compares key fields and detects conflicts.
</VSCode.Cell>
<VSCode.Cell language="python">
from pathlib import Path
import re
import zipfile

base = Path(r"d:\My data\Applying\portfolio")
portfolio_path = base / "index.html"
resume_docx = base / "Varadharajan_Resume.docx"
resume_pdf = base / "resume cv" / "Varadharajan Resume.pdf"

with portfolio_path.open("r", encoding="utf-8", errors="ignore") as f:
    portfolio_text = f.read()

print("Portfolio loaded", len(portfolio_text), "characters")
print("Resume DOCX exists:", resume_docx.exists())
print("Resume PDF exists:", resume_pdf.exists())

resume_text = ""
if resume_docx.exists():
    try:
        with zipfile.ZipFile(resume_docx, "r") as z:
            doc_xml = z.read("word/document.xml").decode("utf-8", errors="ignore")
            resume_text = re.sub(r"<.*?>", " ", doc_xml)
            resume_text = re.sub(r"\s+", " ", resume_text).strip()
            print("Extracted DOCX text length:", len(resume_text))
    except Exception as e:
        print("DOCX extraction failed:", e)
else:
    print("DOCX resume file not found.")

keywords = ["18 years", "19 years", "MBA", "Full Stack", "Corporate Trainer", "Magendhiran", "Infosys", "EduBridge", "Client 4"]
for kw in keywords:
    if kw.lower() in portfolio_text.lower():
        print("Portfolio contains:", kw)

resume_preview = resume_text[:1500]
print("---\nResume preview:\n", resume_preview)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Clean and Normalize Text

Normalize extracted text to prepare for field comparison.
</VSCode.Cell>
<VSCode.Cell language="python">
import string

def normalize_text(text):
    text = text.replace("’", "'").replace("“", '"').replace("”", '"')
    text = text.replace("–", "-").replace("—", '-')
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s\-\.@&]", "", text)
    return re.sub(r"\s+", " ", text).strip()

portfolio_norm = normalize_text(portfolio_text)
resume_norm = normalize_text(resume_text)
print("Normalized portfolio length", len(portfolio_norm))
print("Normalized resume length", len(resume_norm))
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Compare Key Fields

Extract key names, titles, dates, and companies from both documents.
</VSCode.Cell>
<VSCode.Cell language="python">
fields = [
    "varadharajan",
    "full stack",
    "trainer",
    "developer",
    "analyst",
    "educator",
    "magendhiran",
    "infosys",
    "edubridge",
    "Client 4",
    "Client 3",
    "Client 2",
    "Client 1",
    "mba",
    "18 years",
    "19 years",
    "master trainer",
    "corporate trainer",
]

def search_terms(text, terms):
    results = {}
    for term in terms:
        results[term] = term in text
    return results

portfolio_matches = search_terms(portfolio_norm, fields)
resume_matches = search_terms(resume_norm, fields)
print("Portfolio matches:")
print({k:v for k,v in portfolio_matches.items() if v})
print("Resume matches:")
print({k:v for k,v in resume_matches.items() if v})
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Detect Conflicts

Compare important numeric or date fields for mismatches.
</VSCode.Cell>
<VSCode.Cell language="python">
conflicts = []

if "18 years" in portfolio_norm and "19 years" in portfolio_norm:
    conflicts.append("Portfolio uses both 18 and 19 years of experience in different sections.")

if resume_norm:
    if "18 years" in resume_norm and "19 years" in portfolio_norm:
        conflicts.append("Resume says 18 years while portfolio hero says 19 years.")
    if "19 years" in resume_norm and "18 years" in portfolio_norm:
        conflicts.append("Resume says 19 years while portfolio about says 18 years.")
else:
    conflicts.append("Resume text could not be extracted reliably.")

print("Conflicts found:")
for c in conflicts:
    print("-", c)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Summary and Guidance

Based on the extracted data, review the conflicts and apply corrections consistently in both portfolio and resume.
</VSCode.Cell>
<VSCode.Cell language="python">
summary = {
    "portfolio_title": "Varadharajan A — Full Stack Trainer & Developer",
    "experience_text": "Portfolio hero: 19 years; About section: 18 years.",
    "core_conflicts": conflicts,
    "suggestions": [
        "Choose one experience total (18 or 19 years) and use it consistently across hero, about, stats, and resume.",
        "Confirm whether the resume file is the final version and make sure the portfolio download link points to the same final resume document.",
        "Ensure titles align: if your resume emphasizes Trainer and Developer, the portfolio should use the same exact role names.",
    ]
}
print(summary)
```
